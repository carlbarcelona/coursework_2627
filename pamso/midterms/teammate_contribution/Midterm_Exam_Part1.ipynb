```python
import numpy as np
import pandas as pd
```

```python
data = {
    "Name": ["John", "James", "Helena", "Sam"],
    "Math": [85, 72, 90, 65],
    "Science": [80, 75, 88, 70],
    "English": [90, 68, 92, 75]
}

df = pd.DataFrame(data)
print(df)
```

```python
df["Total"] = df["Math"] + df["Science"] + df["English"]
print(df)
```

```python
df["Average"] = df[["Math", "Science", "English"]].mean(axis=1).round(2)
print(df)
```

```python
def get_grade(average):
    if average >= 80:
        return "A"
    elif average >= 70:
        return "B"
    else:
        return "C"

df["Grade"] = df["Average"].apply(get_grade)
print(df)
```
